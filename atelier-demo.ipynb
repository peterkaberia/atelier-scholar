```python
import os
from getpass import getpass

print("Enter your OpenAI API Key:")
os.environ["OPENAI_API_KEY"] = getpass()

print("Enter your OpenAlex API Key (Optional, press enter to skip):")
os.environ["OPENALEX_API_KEY"] = getpass()

# Write these to the .env file Atelier expects
with open('.env', 'w') as f:
    f.write(f"OPENAI_API_KEY={os.environ['OPENAI_API_KEY']}\n")
    f.write(f"OPENALEX_API_KEY={os.environ['OPENALEX_API_KEY']}\n")
```
